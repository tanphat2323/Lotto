````python
import os
import agentops
from openai import OpenAI
from dotenv import load_dotenv

load_dotenv()

# API keys
AGENTOPS_API_KEY = os.getenv("AGENTOPS_API_KEY")
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

# Khởi tạo
agentops.init(api_key=AGENTOPS_API_KEY, default_tags=["lotto-mvp-generator"])
client = OpenAI(api_key=OPENAI_API_KEY)

SYSTEM_PROMPT = """
You are an expert ML engineer. Generate complete Python code for a Lotto 5/35 predictive system MVP.

Dataset: date, id, result1-5 (1-35), db (1-12). NO jackpot/rolldown data.

Requirements:
1. src/ingest/loader.py - Load CSV, validate, save to SQLite
2. src/features/pipeline.py - Feature engineering (sum, parity, gaps, rolling stats, etc.)
3. src/model/builder.py - Bidirectional LSTM -> Attention -> two heads (main 35-sigmoid, special 12-softmax)
4. src/ga/optimizer.py - Genetic algorithm wheel optimizer using LSTM probabilities
5. src/rl/agent.py - PPO agent deciding {skip, buy N, buy GA-wheel} with monetary reward
6. src/backtest/runner.py - Compare strategies: random, top-k, LSTM, GA, RL
7. Dockerfile, requirements.txt, README.md with quickstart

Output full code for each file (100-150 lines each). Use Vietnamese comments where helpful.
"""

@agentops.track_agent(name="lotto-mvp-code-generator")
def generate_mvp():
    print("🤖 Generating Lotto 5/35 MVP skeleton...")

    response = client.chat.completions.create(
        model="gpt-4o",
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": "Generate complete MVP code for Lotto 5/35 system. Include all 7 files with full implementation."}
        ],
        temperature=0.3,
        max_tokens=4000
    )

    return response.choices[0].message.content

if __name__ == "__main__":
    try:
        mvp_code = generate_mvp()

        # Lưu output
        with open("lotto_mvp_generated.md", "w", encoding="utf-8") as f:
            f.write(mvp_code)
        print("\n✅ MVP code generated and saved to lotto_mvp_generated.md")
        agentops.end_session("Success")

    except Exception as e:
        print(f"❌ Error during MVP code generation: {e}")
        agentops.end_session("Fail")
````
